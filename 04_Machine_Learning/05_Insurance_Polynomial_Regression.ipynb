```python
df = pd.read_csv("insurance.csv")
```

```python
df.drop(columns=["friends","enemies","chips","fries","ramen","toportyu"],inplace=True)
```

```python
df.bmi = df.bmi.fillna(df.bmi.mean())
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder  =  OneHotEncoder()
```

```python
encoded_categories = encoder.fit_transform(df.select_dtypes('O'))
```

```python
encoder.get_feature_names_out()
```

```python
encoded_categories = encoded_categories.toarray()
```

```python
type(encoded_categories)
```

```python
encoded_categories = pd.DataFrame(data=encoded_categories, columns=encoder.get_feature_names_out())
```

```python
encoded_categories
```

```python
numerical_data = df.select_dtypes('number')
numerical_data
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
```

```python
scaled_numerical_data = scaler.fit_transform(numerical_data)
```

```python
scaled_numerical_data
```

```python
scaler.get_feature_names_out()
```

```python
numerical_data = pd.DataFrame(data=scaled_numerical_data, columns=scaler.get_feature_names_out())
```

```python
numerical_data
```

```python
df = pd.concat([numerical_data,encoded_categories],axis=1)
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score , mean_squared_error
```

```python
X = df.drop('charges',axis=1)
y = df.charges
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred_train = model.predict(X_train)
```

```python
mean_squared_error(y_train,y_pred_train)
```

```python
r2_score(y_train,y_pred_train)
```

```python
y_pred_test = model.predict(X_test)
```

```python
mean_squared_error(y_test,y_pred_test)
```

```python
r2_score(y_test,y_pred_test)
```

# Applying Polynomial Regression

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(degree=2, include_bias=True)
X_poly = poly.fit_transform(X)
```

```python
X = pd.DataFrame(data=X_poly, columns=poly.get_feature_names_out())
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
y_pred_train = lr.predict(X_train)
```

```python
r2_score(y_train,y_pred_train)
```

```python
mean_squared_error(y_train,y_pred_train)
```

```python
y_pred_test = lr.predict(X_test)
```

```python
r2_score(y_test,y_pred_test)
```

```python
mean_squared_error(y_test,y_pred_test)
```
