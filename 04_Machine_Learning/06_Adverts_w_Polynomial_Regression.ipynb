```python
import pandas as pd
```

```python
df = pd.read_excel("advertising.xlsx")
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
x = df.drop(columns="Sales")
```

```python
x
```

```python
scaled_df = scaler.fit_transform(x)
```

```python
scaled_df
```

```python
scaled_df = pd.DataFrame(data=scaled_df, columns=scaler.get_feature_names_out())
```

```python
scaled_df
```

```python
df = pd.concat([scaled_df,df.Sales],axis=1)
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score , mean_squared_error
```

```python
X = df.drop('Sales',axis=1)
y = df.Sales
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred_train = model.predict(X_train)
```

```python
mean_squared_error(y_train,y_pred_train)
```

```python
r2_score(y_train,y_pred_train)
```

```python
y_pred_test = model.predict(X_test)
```

```python
mean_squared_error(y_test,y_pred_test)
```

```python
r2_score(y_test,y_pred_test)
```

# Applying Polynomial Regression 

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(degree=2, include_bias=True)
X_poly = poly.fit_transform(X)
```

```python
X = pd.DataFrame(data=X_poly, columns=poly.get_feature_names_out())
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
y_pred_train = lr.predict(X_train)
```

```python
r2_score(y_train,y_pred_train)
```

```python
mean_squared_error(y_train,y_pred_train)
```

```python
y_pred_test = lr.predict(X_test)
```

```python
r2_score(y_test,y_pred_test)
```

```python
mean_squared_error(y_test,y_pred_test)
```
